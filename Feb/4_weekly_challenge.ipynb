```python
# 1. 가상의 학생 성적 데이터를 활용한 데이터 전처리 과정 구현해보세요.

from IPython.display import display
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

students = [chr(i) for i in range(ord('A'), ord('Z')+1)]
num_students = len(students)
scores = np.random.randint(30, 101, num_students).astype(float)
nan_mask = np.random.rand(num_students) < 0.2
scores[nan_mask] = np.nan

data = {
    '학생': students,
    '점수': scores
}

df = pd.DataFrame(data)

print(df.info())
print(df.isna().sum())

for i in df.select_dtypes(include=np.number).columns:
    if df[i].isna().sum() > 0:
        df.loc[:, i] = df[i].fillna(df[i].median())

for i in df.select_dtypes(include='object').columns:
    if df[i].isna().sum() > 0:
        df.loc[:, i] = df[i].sort_values(ascending=False).iloc[0]

for i in df.select_dtypes(include=np.number).columns:
    q1 = df[i].quantile(0.25)
    q3 = df[i].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    df = df[(df[i] >= lower_bound) & (df[i] <= upper_bound)].copy()

scaler = MinMaxScaler()
for i in df.select_dtypes(include=np.number).columns:
    df.loc[:, i] = scaler.fit_transform(df[[i]])  
    
display(df)

X = df['학생']
y = df['점수']

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=102)

print(X_train.size,X_test.size,y_train.size,y_test.size)
```

```python
# 4. OR, NOT 연산을 모델링해보세요.

import numpy as np

class AndModel:
    def __init__(self):
        self.weights = np.random.rand(2)
        self.bias = np.random.rand(1)

    def train(self):
        learning_rate = 0.1
        epochs = 20
        inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
        outputs = np.array([0, 0, 0, 1])        
        for epoch in range(epochs):
            for i in range(len(inputs)):
                total_input = np.dot(inputs[i], self.weights) + self.bias
                prediction = self.step_function(total_input)
                error = outputs[i] - prediction
                self.weights += learning_rate * error * inputs[i]
                self.bias += learning_rate * error

    def step_function(self, x):
        return 1 if x >= 0 else 0
    
    def predict(self, input_data):
        total_input = np.dot(input_data, self.weights) + self.bias
        return self.step_function(total_input)

class ORModel:
    def __init__(self):
        self.weights = np.random.rand(2)
        self.bias = np.random.rand(1)

    def train(self):
        learning_rate = 0.1
        epochs = 20
        inputs = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
        outputs = np.array([0, 1, 1, 1])        
        for epoch in range(epochs):
            for i in range(len(inputs)):
                total_input = np.dot(inputs[i], self.weights) + self.bias
                prediction = self.step_function(total_input)
                error = outputs[i] - prediction
                self.weights += learning_rate * error * inputs[i]
                self.bias += learning_rate * error

    def step_function(self, x):
        return 1 if x >= 0 else 0
    
    def predict(self, input_data):
        total_input = np.dot(input_data, self.weights) + self.bias
        return self.step_function(total_input)    

class NOTModel:
    def __init__(self):
        self.weights = np.random.rand(1)
        self.bias = np.random.rand(1)

    def train(self):
        learning_rate = 0.1
        epochs = 20
        inputs = np.array([[0],[1]])
        outputs = np.array([1, 0])        
        for epoch in range(epochs):
            for i in range(len(inputs)):
                total_input = np.dot(inputs[i], self.weights) + self.bias
                prediction = self.step_function(total_input)
                error = outputs[i] - prediction
                self.weights += learning_rate * error * inputs[i]
                self.bias += learning_rate * error

    def step_function(self, x):
        return 1 if x >= 0 else 0
    
    def predict(self, input_data):
        total_input = np.dot(input_data, self.weights) + self.bias
        return self.step_function(total_input)    

class XORModel:
    def __init__(self, and_model, or_model, not_model):
        self.and_model = and_model
        self.or_model = or_model
        self.not_model = not_model

    def predict(self, input_data):
        A, B = input_data
        and_result = self.and_model.predict(np.array([A, B]))
        not_result = self.not_model.predict(np.array([and_result]))
        or_result = self.or_model.predict(np.array([A, B]))
        xor_result = self.and_model.predict(np.array([or_result, not_result]))

        return xor_result

and_model = AndModel()
and_model.train()

or_model = ORModel()
or_model.train()

not_model = NOTModel()
not_model.train()

xor_model = XORModel(and_model, or_model, not_model)

print('AND 연산')
print(f"[0,0] : {and_model.predict([0, 0])}")
print(f"[0,1] : {and_model.predict([0, 1])}")
print(f"[1,0] : {and_model.predict([1, 0])}")
print(f"[1,1] : {and_model.predict([1, 1])}")

print('OR 연산')
print(f"[0,0] : {or_model.predict([0, 0])}")
print(f"[0,1] : {or_model.predict([0, 1])}")
print(f"[1,0] : {or_model.predict([1, 0])}")
print(f"[1,1] : {or_model.predict([1, 1])}")

print('NOT 연산')
print(f"[0] : {not_model.predict([0])}")
print(f"[1] : {not_model.predict([1])}")
```
