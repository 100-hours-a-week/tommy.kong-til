1. 비선형 활성화 함수 3가지(시그모이드, 하이퍼볼릭 탄젠트, 렐루)를 정의하고 -10부터 10까지의 값 100개를 생성한 뒤 입력값으로 사용하여 비선형 활성화 함수를 적용한 값 출력해보세요.


```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(X: np.ndarray) -> np.ndarray:
    return 1 / (1 + np.exp(-X))

def relu(X: np.ndarray) -> np.ndarray:
    return np.maximum(0, X)

def hyperbolic_tangent(X: np.ndarray) -> np.ndarray:
    return np.tanh(X)

X = np.arange(-10, 10, 0.1)
Y_sigmoid = sigmoid(X)
Y_relu = relu(X)
Y_hyperbolic_tangent = hyperbolic_tangent(X)

print(Y_sigmoid)
print('---'*20)
print(Y_relu)
print('---'*20)
print(Y_hyperbolic_tangent)
print('---'*20)

plt.plot(X, Y_sigmoid, label='Sigmoid')
plt.plot(X, Y_relu, label='ReLU')
plt.plot(X, Y_hyperbolic_tangent, label='Hyperbolic Tangent')
plt.legend()
plt.grid(True)
plt.show()
```
